```python
import os
if not os.path.exists("imgs"):
    os.makedirs("imgs")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dias = np.arange(1, 11)
temperaturas = [20, 22, 19, 23, 25, 24, 26, 27, 28, 30]

plt.plot(dias, temperaturas, marker='o', linestyle='-', label='Temperatura')
plt.title("Temperatura a lo largo de 10 días")
plt.xlabel("Días")
plt.ylabel("Temperatura (°C)")
plt.legend()

plt.savefig('imgs/line_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

Este gráfico muestra cómo varía la temperatura a lo largo de los días. Se utiliza un line plot porque permite visualizar tendencias en el tiempo de forma clara.

```python
materias = ['Math', 'Physics', 'Programming', 'English']
horas = [5, 3, 6, 2]

plt.bar(materias, horas)
plt.title("Horas de estudio por materia")
plt.xlabel("Materia")
plt.ylabel("Horas")

plt.savefig('imgs/bar_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

Este gráfico compara las horas dedicadas a cada materia. Se usa un gráfico de barras porque es ideal para comparar cantidades entre categorías.

```python
notas = np.random.normal(6, 1.5, 100)

plt.hist(notas, bins=10)
plt.title("Distribución de notas")
plt.xlabel("Nota")
plt.ylabel("Frecuencia")

plt.savefig('imgs/histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```

Este histograma muestra cómo se distribuyen las notas. Se usa porque permite ver la frecuencia de valores en intervalos.

```python
horas = np.random.randint(1, 10, 50)
notas = horas + np.random.normal(0, 1, 50)

plt.scatter(horas, notas)
plt.title("Relación entre horas de estudio y nota")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")

plt.savefig('imgs/histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```

Este gráfico muestra la relación entre horas de estudio y rendimiento. Se utiliza scatter plot porque permite ver correlaciones entre variables.

```python
import pandas as pd

data = pd.DataFrame({
    'Horas': np.random.randint(1, 10, 50),
    'Sueño': np.random.randint(4, 10, 50),
    'Nota': np.random.randint(1, 10, 50)
})

sns.heatmap(data.corr(), annot=True)
plt.title("Correlación entre variables")

plt.savefig('imgs/custom_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

Este heatmap muestra la correlación entre variables. Se utiliza porque permite identificar relaciones positivas o negativas de forma visual.
